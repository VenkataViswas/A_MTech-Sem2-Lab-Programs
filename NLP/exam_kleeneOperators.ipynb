# 1. *

```python
import re
pattern = r"baa*$"
word ="baaaa"
if re.match(pattern,word):
  print("sheep language")
else :
  print("not sheep language")
```

2. \+

```python
import re
pattern = r"wo+w$"
word ="wooow"
if re.match(pattern,word):
  print("pattern matched")
else :
  print("pattern not matched ")
```

3. ?

```python
import re
pattern = r"books?$"
word ="book"
if re.match(pattern,word):
  print("pattern matched")
else :
  print("pattern not matched ")
```

4.ngrams

```python
from nltk import ngrams
sentance = "i love nlp".split()
n = 2
ng = ngrams(sentance,n)
for i in ng :
  print(i)
```
